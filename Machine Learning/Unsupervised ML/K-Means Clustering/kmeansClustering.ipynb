## K-means Clustering


```python
import seaborn as sns
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(n_samples=1000, n_features=2, centers=4, random_state=42)
```

```python
sns.scatterplot(x=X[:, 0], y=X[:, 1])
```

```python
# k means clustering

from sklearn.cluster import KMeans
```

```python
K = 4

kmeans = KMeans(n_clusters=K, random_state=42)
```

```python
labels = kmeans.fit_predict(X)
# labels = cluster number
```

```python
sns.scatterplot(x=X[:, 0], y=X[:, 1], c=labels)
```

## How to choose our k value - elbow; silhouette score


```python
wcss = []
for k in range(1, 21):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1, 21), y=wcss, marker="o")
```

```python
# kneed module
from kneed import KneeLocator
```

```python
kneeLocater = KneeLocator(range(1, 21), wcss, curve="convex", direction="decreasing")
```

```python
print("Optimal k =", kneeLocater.elbow)
```

## Silhouette Score


```python
from sklearn.metrics import silhouette_score
```

```python
ss = []

for k in range(2, 21):
    kmeans = KMeans(n_clusters=k)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)

    ss.append(score)
```

```python
sns.lineplot(x=range(2, 21), y=ss, marker="o")
```
