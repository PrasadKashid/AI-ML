## Classifier


```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.tree import DecisionTreeClassifier
```

```python
X, y = make_classification(
    n_samples=500, n_features=20, n_informative=10, n_redundant=2, random_state=42
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.2
)
```

```python
from sklearn.ensemble import AdaBoostClassifier

# DT stump

base_model = DecisionTreeClassifier(max_depth=1)

abc = AdaBoostClassifier(estimator=base_model, n_estimators=100, random_state=42)
```

```python
abc.fit(X_train, y_train)
```

```python
y_pred = abc.predict(X_test)
```

```python
print("Accuracy Score : ", accuracy_score(y_test, y_pred))
print("Classification Report : \n", classification_report(y_test, y_pred))
```

```python
from sklearn.datasets import make_regression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score
```

```python
X, y = make_regression(
    n_samples=1000, n_features=20, n_informative=10, noise=20, random_state=42
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.2
)
```

```python
base_model = DecisionTreeRegressor(max_depth=1)

abr = AdaBoostRegressor(n_estimators=100, estimator=base_model, random_state=42)
```

```python
abr.fit(X_train, y_train)
```

```python
y_pred = abr.predict(X_test)
```

```python
print("r2 Score",r2_score(y_test, y_pred))
```
